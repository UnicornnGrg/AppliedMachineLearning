```python
import pandas as pd

# Read CSV
df_raw = pd.read_csv("csv_pny/psam_p36.csv", dtype=str)
df_raw
```

```python
# Create Subset
cols = [
    # IDs / meta (not features later, just for reference)
    "SERIALNO", "SPORDER", "PWGTP",

    # target + coverage detail (HINS*, PRIVCOV/PUBCOV used only for sanity, not as features)
    "HICOV", "HINS1", "HINS2", "HINS3", "HINS4", "HINS5", "HINS6", "HINS7",
    "PRIVCOV", "PUBCOV",
    
    # GROUP 1: Income & Affordability
    "WAGP", "POVPIP", "SEMP", "INTP", "RETP", "SSP", "SSIP", "PAP", #"PINCP",

    # GROUP 2: Employment Quality
    "ESR", "COW", "WKL", "OCCP", "INDP",

    # GROUP 3: Demographics & Life Stage
    "AGEP", "SEX", "MAR", "MIG", "NATIVITY", "RAC1P", "HISP", #"CIT",

    # GROUP 4: Education
    "SCHL", "SCH",

    # GROUP 5: Health & Disability
    "DIS", "DEAR", "DEYE", "DPHY",

    # GROUP 6: Geography
    "STATE", "PUMA", "REGION", "DIVISION",
]

df_subset = df_raw[cols].copy()

df_subset.head(20)
```

```python
na_summary = (
    df_subset.isna()
    .sum()
    .to_frame("na_count")
    .assign(na_pct=lambda x: x["na_count"] / len(df_subset))
)

print(na_summary)
```

```python
# Identify columns with at least one NA
cols_with_na = na_summary.query("na_count > 0").index.tolist()

# Create subset with only those columns
df_with_na = df_subset[cols_with_na].copy()

df_with_na.head(50)

df_with_na.to_csv("df_with_na_columns.csv", index=False)
```
